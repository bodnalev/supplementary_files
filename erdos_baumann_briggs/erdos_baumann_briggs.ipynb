Problems of Erdos and Baumann-Briggs
============

Write $K_k, E_k, C_k$ for the complete, empty and cycle graphs on $k$ vertices respectively. Let $er_s(F)$ denote the minimum density of $K_s$ in limit graphs without induced copies of $F$. 

1. The first section calculates $er_8(E_3)$, $er_{s}(E_3, C_5)$ for $s \in \{4, 5, 6\}$, and $er_{s}(E_4, C_5+K_1)$ for $s \in \{4, 5\}$.
2. In the second section, we perform the test for perfect stability on $er_8(3)$. Additionally, we extract infromation from the other certificates to prove structural results about the optimum.
3. Finally, we include the verifier section, checking that all certificate provides the claimed bound.

Calculation for $er_8(E_3)$
------------

```python
s = 8
target_size = 8


G = GraphTheory
G.printlevel(1)
G.reset()
E3 = G(3)
G.exclude(E3)

clebsch_edges = []
def sbin(n):
    return sum(map(int,"{0:b}".format(n)))
for xx in range(2**4):
    xxp = xx<<1 if sbin(xx)%2==0 else (xx<<1) + 1
    for yy in range(xx, 2**4):
        yyp = yy<<1 if sbin(yy)%2==0 else (yy<<1) + 1
        hd = sbin(xxp ^^ yyp)
        if hd!=4:
            clebsch_edges.append([xx, yy])
print(f"L has edges: {" ".join(map(str, clebsch_edges))}")
cons = G.blowup_construction(target_size, 16, edges=clebsch_edges)

target = G(s, edges=list(itertools.combinations(range(s), 2)))
lb = cons.density(target)
print(f"The density of K_8 in L is {lb}~{lb.n()}")

G.solve_sdp(
    target, target_size, file="certificates/er83_problem",
    construction=cons, maximize=False, 
    solver="/usr/local/bin/sdpa_qd"
)
ub = G.round_solution(
    "certificates/er83_problem", "certificates/er83_cert", 
    denom=2**20, kernel_denom=2**15, kernel_threshold=1e-5
)
print(f"The found lower bound is {ub}~{ub.n()}")
```

Getting structural results
===========

Perfect stability for $er_8(E_3)$
---------

```python
from fractions import Fraction
from sage.algebras.combinatorial_theory import _unflatten_matrix
import pickle


G = GraphTheory
G.reset()
E3 = G(3)
G.exclude(E3)

clebsch_edges = []
def sbin(n):
    return sum(map(int,"{0:b}".format(n)))
for xx in range(2**4):
    xxp = xx<<1 if sbin(xx)%2==0 else (xx<<1) + 1
    for yy in range(xx, 2**4):
        yyp = yy<<1 if sbin(yy)%2==0 else (yy<<1) + 1
        hd = sbin(xxp ^^ yyp)
        if hd!=4:
            clebsch_edges.append([xx, yy])

def to_sage(dim, data):
    if dim==0:
        if isinstance(data, Fraction):
            return QQ(data)
        if isinstance(data, float):
            return RR(data)
        return data
    return [to_sage(dim-1, xx) for xx in data]

cert_file = "certificates/er83_cert.pickle"

print("checking stability for " + cert_file)
G.printlevel(0)
with open(cert_file, "rb") as f:
    certificate = pickle.load(f)
target_size = certificate["target size"]
original_bound = to_sage(0, certificate["result"])
print("Target size is ", target_size)

construction = G.blowup_construction(target_size, 16, edges=clebsch_edges)
cvals = construction.values()
svals = to_sage(1, certificate["slack vector"])
assert(len(cvals)==len(svals))
correct_slacks = True
for ii in range(len(svals)):
    if svals[ii]==0 and cvals[ii]==0:
        correct_slacks = False
if not correct_slacks:
    print("Compl slackness fails, the following flags violate it:")
    for ii,ff in enumerate(G.generate(target_size)):
        if svals[ii]==0 and cvals[ii]==0:
            print(ff)
else:
    print("Compl slackness holds!")

goal_type = G._get_relevant_ftypes(8)[31][1]
print(f"Checking nullity for {goal_type}")

for ii, kk in enumerate(certificate["typed flags"].keys()):
    if ii==31:
        mat = matrix(to_sage(2, _unflatten_matrix(certificate["X matrices"][ii])[0]))
        mn = mat.nullity()
        print(f"Nullity is {mn} for type {kk}")
```

Examining the slack values for the remaining problems
---------------

```python
from fractions import Fraction
from sage.algebras.combinatorial_theory import _unflatten_matrix
import pickle
import itertools

def to_sage(dim, data):
    if dim==0:
        if isinstance(data, Fraction):
            return QQ(data)
        if isinstance(data, float):
            return RR(data)
        return data
    return [to_sage(dim-1, xx) for xx in data]

G = GraphTheory
G.printlevel(0)
G.reset()
E3 = G(3)
C5 = G(5, edges=[[0, 1], [1, 2], [2, 3], [3, 4], [4, 0]])
G.exclude([E3, C5])

for s in [4, 5, 6]:
    cert_file = f"certificates/BB_e3c5_s{s}_cert.pickle"
    print("\n\nchecking file " + cert_file)
    with open(cert_file, "rb") as f:
        certificate = pickle.load(f)
    target_size = certificate["target size"]
    test_graph = G(target_size, edges=[ee for ee in itertools.combinations(range(target_size), 2) if ee!=(0, 1)])
    svals = to_sage(1, certificate["slack vector"])
    test_index = G.generate(target_size).index(test_graph)
    print(f"Slack value is {svals[test_index].n()} for the graph {test_graph}")

G.reset()
E4 = G(4)
C5pK1 = G(6, edges=[[0, 1], [1, 2], [2, 3], [3, 4], [4, 0]])
G.exclude([E4, C5pK1])

cert_file = "certificates/BB_e4c5k1_s4_cert.pickle"
print("\n\nchecking file " + cert_file)
with open(cert_file, "rb") as f:
    certificate = pickle.load(f)
target_size = certificate["target size"]
test_graph = G(target_size, edges=[ee for ee in itertools.combinations(range(target_size), 2) if ee!=(0, 1)])
svals = to_sage(1, certificate["slack vector"])
test_index = G.generate(target_size).index(test_graph)
print(f"Slack value is {svals[test_index].n()} for the graph {test_graph}")
```
